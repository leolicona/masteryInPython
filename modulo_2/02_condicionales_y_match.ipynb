# Condicionales (If, Elif, Else) y Match


## 1. Condicionales If / Else
El `if` (si ocurre esto) permite que tu código tome distintos caminos. El código dentro de un `if` **debe estar indentado** (llevar espacios al principio).


```python
clima = "lluvia"

if clima == "lluvia":
    print("Lleva paraguas")
elif clima == "nublado":
    print("Lleva chamarra")
else:
    print("Día soleado, ponte lentes")
```

## 2. La sentencia Match (Python 3.10+)
Cuando tienes muchísimos `elif` seguidos comprobando la misma variable, es mejor usar `match` (similar al switch de otros lenguajes).


```python
dia = "Lunes"

match dia:
    case "Lunes":
        print("Inicio de semana")
    case "Viernes":
        print("¡Por fin!")
    case _:
        print("Día normal") # El _ significa "cualquier otro caso"
```
