# Operadores Básicos y Lógicos


## 1. Operadores Aritméticos y de Asignación
Ya vimos la suma y la resta. A veces queremos actualizar una variable usando su propio valor (como sumar puntos en un juego).


```python
puntos = 10
puntos = puntos + 5   # Forma tradicional
print(puntos)

puntos += 5          # Forma corta (asignación con adición)
print(puntos)
```

## 2. Operadores Lógicos (and, or, not)
Para tomar decisiones complejas, necesitamos combinar varias preguntas de verdadero/falso.


```python
tiene_llave = True
es_de_dia = False

print("Puede entrar si tiene llave Y es de día:", tiene_llave and es_de_dia)
print("Puede entrar si tiene llave O es de día:", tiene_llave or es_de_dia)
print("¿NO es de día?", not es_de_dia)
```
