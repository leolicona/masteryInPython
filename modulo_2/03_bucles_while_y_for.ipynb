# Bucles While y For


## 1. Bucle While
Un `while` repite un bloque de código **mientras** una condición sea Verdadera. ¡Cuidado con crear bucles infinitos!


```python
contador = 3
while contador > 0:
    print(f"Despegue en {contador}...")
    contador -= 1
print("¡Fuego!")
```

## 2. Bucle For
El `for` se usa para recorrer colecciones de elementos (como las letras de un string, o un rango de números).


```python
palabra = "Python"
for letra in palabra:
    print(f"Dame una {letra}!")

print("---")
# Usando range(inicio, fin)
for i in range(1, 4):
    print(f"Intento número {i}")
```

## 3. Break y Continue
`break` rompe el bucle por completo. `continue` salta directamente a la siguiente repetición.


```python
for numero in range(1, 10):
    if numero == 3:
        continue  # Se salta el 3
    if numero == 6:
        break     # Se detiene al llegar a 6
    print(numero)
```
