# Introducción a APIs y Requests


## 1. ¿Qué es una API?
Una API (Application Programming Interface) es como un mesero en un restaurante. Tú (el cliente) le das tu orden al mesero (la API), este la lleva a la cocina (el servidor), y luego te trae la comida (los datos).

En Python, la librería más popular para ser el cliente es `requests`.


```python
import requests

# Vamos a pedir datos a una API pública y gratuita
respuesta = requests.get("https://pokeapi.co/api/v2/pokemon/pikachu")

if respuesta.status_code == 200:
    datos = respuesta.json()
    print(f"Nombre: {datos['name']}")
    print(f"Peso: {datos['weight']}")
else:
    print("Error al conectar con la API")
```
