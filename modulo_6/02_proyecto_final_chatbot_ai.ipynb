# Proyecto Final: Chatbot AI


## 🚀 Proyecto Final: Chatbot con Gemini AI
Vamos a unir funciones, bucles `while`, manejo de errores (`try/except`) y llamadas a APIs para construir un asistente que hable contigo usando la API de Google Gemini.

*Nota: Para que este código funcione, necesitarás generar tu propia API KEY gratuita en Google AI Studio (aistudio.google.com).*


```python
import requests

# 1. Configuración
API_KEY = "PON_TU_API_KEY_AQUI"
URL = f"https://generativelanguage.googleapis.com/v1beta/models/gemini-pro:generateContent?key={API_KEY}"

def preguntar_a_gemini(texto):
    # 2. Preparamos los datos en formato JSON como lo pide la API
    payload = {"contents": [{"parts": [{"text": texto}]}]}
    headers = {"Content-Type": "application/json"}
    
    # 3. Hacemso la petición POST protegida por try/except
    try:
        respuesta = requests.post(URL, json=payload, headers=headers)
        if respuesta.status_code == 200:
            datos = respuesta.json()
            return datos["candidates"][0]["content"]["parts"][0]["text"]
        else:
            return "Error en la API."
    except Exception as e:
        return f"Error de conexión: {e}"

# 4. El bucle infinito interactivo
print("🤖 Chatbot iniciado. Escribe 'salir' para terminar.")
while True:
    usuario = input("Tú: ")
    if usuario.lower() == "salir":
        print("🤖 Adiós!")
        break
    
    if API_KEY != "PON_TU_API_KEY_AQUI":
        respuesta_bot = preguntar_a_gemini(usuario)
        print(f"🤖 Bot: {respuesta_bot}")
    else:
        print("🤖 Bot: (Por favor configura tu API KEY en el código primero)")
        break
```
