# Manejo de Errores


## 1. Try / Except
Los errores en programación (excepciones) detienen tu código abruptamente. Con `try` y `except` puedes atrapar esos errores y manejarlos con gracia.


```python
try:
    resultado = 10 / 0
except ZeroDivisionError:
    print("¡Error! No puedes dividir entre cero.")
except Exception as e:
    print(f"Ocurrió un error inesperado: {e}")
print("El programa sigue funcionando...")
```

## 2. Finally
El bloque `finally` se ejecuta **siempre**, haya ocurrido un error o no. Es útil para cerrar archivos o conexiones a bases de datos.


```python
try:
    numero = int("10")
    print("El número es", numero)
except ValueError:
    print("Eso no es un número")
finally:
    print("Operación de conversión terminada.")
```
