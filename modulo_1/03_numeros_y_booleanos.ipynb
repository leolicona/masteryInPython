# Números y Booleanos


## 1. Operaciones Matemáticas Básicas
Python es esencialmente una calculadora muy potente. Puedes usar variables numéricas o números directamente.


```python
a = 10
b = 3

print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicación:", a * b)
print("División:", a / b)      # Siempre devuelve un float
print("División entera:", a // b) # Devuelve un int, descarta decimales
print("Módulo:", a % b)        # Devuelve el resto de la división (muy útil para saber si un número es par)
print("Exponente:", a ** b)    # a elevado a la b (10 al cubo)
```

## 2. Booleanos: Cierto o Falso
Los booleanos (`bool`) son un tipo de dato que solo puede tener dos valores: `True` (Verdadero) o `False` (Falso). Nota que la primera letra siempre va en mayúscula.

Son la base de la toma de decisiones en programación. Normalmente los obtenemos al comparar cosas.


```python
es_mayor = 10 > 5
print("¿10 es mayor que 5?", es_mayor)

print("¿5 es igual a 5?", 5 == 5)    # Usamos == para comparar, = es para asignar
print("¿3 es diferente de 4?", 3 != 4)
print("¿10 es menor o igual a 10?", 10 <= 10)
```

## 3. Conversión de Tipos (Casting)
A veces necesitas transformar un número en texto, o texto en número. Esto se llama *casting*.


```python
numero_texto = "100"
# print(numero_texto + 50)  # Esto daría un error, no puedes sumar texto y número directamente.

numero_real = int(numero_texto)
print("Ahora sí:", numero_real + 50)

# Puedes convertir a booleano. Casi todo es True, excepto el 0, las listas vacías o el None.
print("bool de 1:", bool(1))
print("bool de 0:", bool(0))
print("bool de texto vacío:", bool(""))
```
