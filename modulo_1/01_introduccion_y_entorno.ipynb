## 1. ¿Por qué Python?

## 3. Comentarios en Python
El código no es solo para que lo lean las máquinas; es para que lo lean los humanos. Los comentarios son notas que dejas en el código y que Python ignora por completo. Son vitales para explicar el *por qué* de tu lógica.


```python
# Esto es un comentario de una sola línea.
print("Esto sí se ejecuta")

"""
Esto es un comentario
de múltiples líneas.
Sirve para explicaciones muy largas.
"""
print("Y esto también se ejecuta")
```
