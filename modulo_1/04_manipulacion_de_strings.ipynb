# Manipulación de Strings


## 1. El poder de los Strings
Los strings en Python son muy flexibles. Si tu texto incluye comillas simples, envuelve el string en comillas dobles (y viceversa). Si necesitas un texto de varias líneas (como un poema), usa tres comillas `'''`.


```python
frase = "El programador dijo: 'Python es genial'."
poema = """
Rosa roja
Cielo azul
Python es chido
Y tú también
"""
print(frase)
print(poema)
```

### F-Strings (Format Strings)
La forma más moderna y limpia de meter variables dentro de un texto en Python es usando f-strings. Pon una `f` antes de las comillas, y mete tus variables en `{}`.


```python
nombre = "Leo"
meta = "Maestro de Python"

print(f"Me llamo {nombre} y voy a ser un {meta}!")
```

## 2. Manipulación: Métodos de Strings
Un string viene con "herramientas" integradas llamadas métodos para manipular su texto.


```python
texto = "  Python es asombroso  "

print("Original:", texto)
print("Mayúsculas:", texto.upper())
print("Minúsculas:", texto.lower())
print("Sin espacios a los lados:", texto.strip())
print("Reemplazando palabras:", texto.replace("asombroso", "increíble"))
```

## 3. Slicing (Rebanando Strings) y Búsqueda
Puedes acceder a partes específicas de un string usando corchetes `[]`. En programación, **¡siempre empezamos a contar desde cero!**


```python
palabra = "Programacion"

print("Primera letra:", palabra[0])
print("Última letra:", palabra[-1])   # Índices negativos cuentan desde el final
print("Las primeras 3 letras:", palabra[0:3]) # [inicio:fin] (el fin no se incluye)
print("Desde la índice 5 en adelante:", palabra[5:])

# También podemos convertir un string en una lista de palabras usando split()
frase = "Manzana, Pera, Plátano"
frutas = frase.split(", ")
print("Frutas separadas:", frutas)
```
