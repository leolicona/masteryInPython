# Variables y Tipos de Datos


## 1. ¿Qué es una Variable?
Imagina que una variable es una caja con una etiqueta donde puedes guardar datos. Cuando necesites ese dato más adelante, simplemente llamas a la etiqueta de la caja.

En Python, no necesitas decirle de qué material es la caja (qué tipo de dato vas a guardar), Python lo adivina por ti. Solo usas el signo `=` para asignar un valor a una variable.


```python
nombre = "Leo"       # Guardamos texto
edad = 25          # Guardamos un número entero
altura = 1.75      # Guardamos un número con decimales

print("Me llamo", nombre, "y tengo", edad, "años.")
```

### Reglas para nombrar variables:
- Deben empezar con una letra o un guion bajo `_`.
- No pueden contener espacios (usa `guion_bajo` conocido como *snake_case*).
- No pueden ser palabras reservadas de Python (como `print`, `if`, `while`).


## 2. Tipos de Datos Básicos
En el ejemplo anterior vimos tres tipos de datos básicos:
1. **Strings (`str`)**: Cadenas de texto. Siempre van entre comillas simples `''` o dobles `""`.
2. **Integers (`int`)**: Números enteros, sin decimales.
3. **Floats (`float`)**: Números con decimales.

Puedes usar la función `type()` para descubrir qué tipo de dato hay en una variable.


```python
misterio = 42
print(type(misterio))

misterio = "Cuarenta y dos"
print(type(misterio)) # ¡Python te permite cambiar el tipo de dato de una variable libremente!
```

## 3. Asignación Múltiple
Python tiene un truco genial: puedes asignar valores a múltiples variables en una sola línea. Esto ahorra espacio y hace el código más limpio.


```python
x, y, z = 10, 20, 30
print(x, y, z)

# También puedes dar el mismo valor a varias variables
a = b = c = 100
print(a, b, c)
```

## 4. El tipo de dato None
A veces necesitas una caja vacía, una variable que exista pero que no tenga nada todavía. Para eso usamos `None`.


```python
resultado_pendiente = None
print(type(resultado_pendiente))
```
