# Listas y Tuplas


## 1. Listas
Una lista es una colección ordenada que puede contener varios elementos. Se definen usando corchetes `[]`.


```python
frutas = ["manzana", "banana", "cereza"]
print("Lista original:", frutas)

# Accediendo a elementos
print("El primer elemento es:", frutas[0])

# Modificando
frutas[1] = "kiwi"

# Agregando al final
frutas.append("naranja")

# Eliminando
frutas.remove("manzana")
print("Lista modificada:", frutas)
```

## 2. Tuplas
Las tuplas son como las listas, pero **no se pueden modificar** una vez creadas (son inmutables). Son más rápidas y seguras si tienes datos fijos. Se definen con paréntesis `()`.


```python
coordenadas = (10.5, 20.3)
print(coordenadas[0])
# coordenadas[0] = 15.0  <-- ¡Esto daría error!
```
