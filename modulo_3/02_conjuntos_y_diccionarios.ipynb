# Conjuntos y Diccionarios


## 1. Conjuntos (Sets)
Un conjunto es una colección desordenada que **no permite elementos duplicados**. Son geniales para eliminar duplicados de una lista de forma rápida. Se usan llaves `{}`.


```python
numeros = {1, 2, 2, 3, 3, 4}
print("Conjunto sin duplicados:", numeros)

# Operaciones de conjuntos
A = {1, 2, 3}
B = {3, 4, 5}
print("Unión:", A | B)           # Todos los elementos
print("Intersección:", A & B)    # Solo los que están en ambos
```

## 2. Diccionarios (Dicts)
Son colecciones de pares `clave: valor`. Es como un diccionario de la vida real: buscas una palabra (clave) y encuentras su significado (valor).


```python
usuario = {
    "nombre": "Leo",
    "edad": 25,
    "es_admin": True
}

# Accediendo
print("Nombre:", usuario["nombre"])

# Modificando / Agregando
usuario["edad"] = 26
usuario["ciudad"] = "CDMX"    # Si no existe, se crea
print("Diccionario actualizado:", usuario)
```
