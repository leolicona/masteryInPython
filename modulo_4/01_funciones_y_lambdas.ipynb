# Funciones y Lambdas


## 1. Funciones Básicas
Las funciones son bloques de código reutilizables. Te ahorran escribir lo mismo una y otra vez. Se definen con la palabra `def`.


```python
def saludar(nombre):
    return f"¡Hola {nombre}, bienvenido!"

# Usando la función
mensaje = saludar("Leo")
print(mensaje)
```

## 2. Argumentos por Defecto
Puedes darle a las funciones valores por defecto, por si el usuario olvida pasarlos.


```python
def hacer_cafe(tipo="Americano"):
    print(f"Preparando un rico {tipo} ☕")

hacer_cafe()            # Usa el default
hacer_cafe("Espresso")  # Lo sobreescribe
```

## 3. Funciones Lambda
Son funciones anónimas (sin nombre) y de una sola línea. Se usan para operaciones matemáticas rápidas o para ordenar listas.


```python
doble = lambda x: x * 2
print("El doble de 5 es:", doble(5))

# Muy útil para ordenar listas de diccionarios
usuarios = [{"nombre": "Ana", "edad": 30}, {"nombre": "Leo", "edad": 25}]
usuarios_ordenados = sorted(usuarios, key=lambda u: u["edad"])
print("Ordenados por edad:", usuarios_ordenados)
```
