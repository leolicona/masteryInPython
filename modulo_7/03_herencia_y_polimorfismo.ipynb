# Herencia y Polimorfismo


## 1. Herencia
La herencia permite crear una clase nueva (hija) que hereda todos los métodos y atributos de otra clase (padre). Usamos `super()` para llamar a los métodos del padre.


```python
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre
    def hacer_sonido(self):
        pass

class Gato(Animal):
    def __init__(self, nombre, vidas=7):
        super().__init__(nombre) # Llama al __init__ de Animal
        self.vidas = vidas
    
    def hacer_sonido(self):
        return "¡Miau!"

michi = Gato("Garfield")
print(f"{michi.nombre} hace {michi.hacer_sonido()} y tiene {michi.vidas} vidas.")
```
