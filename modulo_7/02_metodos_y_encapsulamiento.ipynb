# Métodos y Encapsulamiento


## 1. Encapsulamiento
Es una buena práctica ocultar los datos internos de una clase para que no se modifiquen accidentalmente desde afuera. En Python, indicamos que una variable es 'privada' poniendo un guion bajo `_` o dos `__` al principio de su nombre.


```python
class CuentaBancaria:
    def __init__(self, titular, saldo_inicial):
        self.titular = titular
        self.__saldo = saldo_inicial  # Variable "privada"
        
    def depositar(self, cantidad):
        if cantidad > 0:
            self.__saldo += cantidad
            return f"Depósito exitoso. Nuevo saldo: {self.__saldo}"
        return "Cantidad inválida"
        
    def ver_saldo(self):
        return f"El saldo de {self.titular} es {self.__saldo}"

cuenta = CuentaBancaria("Leo", 1000)
print(cuenta.depositar(500))
# print(cuenta.__saldo)  # Esto daría un error, __saldo está protegido desde el exterior.
print(cuenta.ver_saldo())
```
