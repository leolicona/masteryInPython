# Clases y Objetos


## 1. El Paradigma de POO
En la Programación Orientada a Objetos, agrupamos **datos** (variables) y **comportamientos** (funciones) dentro de una misma estructura llamada **Clase**. Puedes pensar en una Clase como un plano arquitectónico, y en un **Objeto** (o instancia) como la casa construida con ese plano.


```python
class Perro:
    # El método __init__ es el "constructor". Se llama automáticamente al crear un nuevo Perro.
    def __init__(self, nombre, raza):
        self.nombre = nombre  # "self" se refiere al objeto específico que estamos creando
        self.raza = raza
        
    def ladrar(self):
        return f"{self.nombre} dice: ¡Guau!"

# Creando objetos (instancias)
mi_perro = Perro("Firulais", "Labrador")
otro_perro = Perro("Rex", "Pastor Alemán")

print(mi_perro.ladrar())
print(f"La raza de Rex es {otro_perro.raza}")
```
